```python
#7
import  numpy as np
asia = df[df["origin"] == "Asia"]

# Take first 7 rows and required columns
X = asia[["vehicle_weight", "model_year"]].head(7).to_numpy()

# Calculate X^T X
XTX = X.T @ X

# Calculate inverse of X^T X
XTX_inv = np.linalg.inv(XTX)

# Create y
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

w = XTX_inv @ X.T @ y

print("Weights:", w)
print("Sum of weights:", w.sum())
```
